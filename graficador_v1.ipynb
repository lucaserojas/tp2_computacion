## Graficador de Mediciones

```python
import matplotlib.pyplot as plt
import numpy as np
import os
import pandas as pd
```

## Directorios y Carga de Archivos

```python
DATA_DIR = "data"
MEASURE_DIR = "mic_behringher_ecm8000"
OUTPUT_DIR = "output"
output_path = os.path.join(OUTPUT_DIR, MEASURE_DIR)
os.makedirs(output_path, exist_ok=True)
files = os.listdir(os.path.join(DATA_DIR, MEASURE_DIR))
```

## Analisis de Datos

```python
path_file = os.path.join(DATA_DIR, MEASURE_DIR, files[0])
df = pd.read_csv(path_file, delimiter="\t", header=1)
df.head()
```

## Graficos de los Resultados

```python
x_ticks = [20, 100, 1000, 10000, 20000]
x_label = ['20', '100', '1k', '10k', '20k'] 
y_ticks = np.arange(-12, 13, 3)
plt.figure(figsize=(10, 6))
plt.plot(df['Frequency (Hz)'], df['Magnitude (dB)'])
plt.title("Respuesta en frecuencia del micrófono Behringer ECM8000 a 0°", fontsize=16)
plt.xscale('log')
plt.xlim(20, 20000)
plt.ylim(-12, 12)
plt.yticks(y_ticks)
plt.xticks(x_ticks, x_label)
plt.xlabel("Frecuencia (Hz)", fontsize=14, fontname= "Arial")
plt.ylabel("Nivel (dB)", fontsize=14, fontname= "Arial")
plt.grid(which='both', linestyle='--', linewidth=0.5)
plt.savefig(os.path.join(output_path, "Behringer_0.png"), dpi=300)
plt.show()
```
